```python
from pathlib import Path

import numpy as np
import pandas as pd

import xarray as xr
from climtools import xgeo
```

```python
# (Assumes XNpyStore class is defined)


```

```python
np_path = Path("./store_ndarray")


arr = np.array([[1.0, 2.0], [3.0, 4.0]])
xgeo.to_xnpy(np_path, arr)

loaded_arr = xgeo.open_xnpy(np_path)
assert np.array_equal(arr, loaded_arr)
loaded_arr
```

```python
pd_path = Path("./store_dataframe")


df = pd.DataFrame(
    {"A": [10, 20, 30], "B": [1.5, 2.5, 3.5]},
    index=pd.Index([0, 1, 2], name="row_id"),
)
xgeo.to_xnpy(pd_path, df)

loaded_df = xgeo.open_xnpy(pd_path)
pd.testing.assert_frame_equal(df, loaded_df)
loaded_df
```

```python
ds = xr.Dataset(
    {
        "temperature": (("x", "y"), np.random.rand(2, 3)),
    },
    coords={
        "x": [10, 20],
        "y": [1, 2, 3],
    },
    attrs={"description": "Test dataset"},
)
xgeo.to_xnpy("./store_dataset", ds)

loaded_ds = xgeo.open_xnpy("./store_dataset")

xr.testing.assert_equal(ds, loaded_ds)

loaded_ds
```
